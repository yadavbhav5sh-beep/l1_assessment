```python
%sql
create table if not exists tpch_dev.bronze.customer;
create table if not exists tpch_dev.bronze.orders;
create table if not exists tpch_dev.bronze.lineitem;
create table if not exists tpch_dev.bronze.nation;
create table if not exists tpch_dev.bronze.part;
create table if not exists tpch_dev.bronze.partsupp;
create table if not exists tpch_dev.bronze.region;
create table if not exists tpch_dev.bronze.supplier;
```

```python
%sql
CREATE OR REPLACE TABLE tpch_dev.bronze.customer AS
SELECT *,
  'samples.tpch.customer' AS _source_file,
  current_timestamp() AS _ingestion_at
FROM samples.tpch.customer;
```

```python
%sql --name bronze_vs_samples
SELECT 'customer' AS table_name, (SELECT COUNT(*) FROM tpch_dev.bronze.customer) AS bronze_count, (SELECT COUNT(*) FROM samples.tpch.customer) AS samples_count
UNION ALL
SELECT 'orders', (SELECT COUNT(*) FROM tpch_dev.bronze.orders), (SELECT COUNT(*) FROM samples.tpch.orders)
UNION ALL
SELECT 'lineitem', (SELECT COUNT(*) FROM tpch_dev.bronze.lineitem), (SELECT COUNT(*) FROM samples.tpch.lineitem)
UNION ALL
SELECT 'nation', (SELECT COUNT(*) FROM tpch_dev.bronze.nation), (SELECT COUNT(*) FROM samples.tpch.nation)
UNION ALL
SELECT 'part', (SELECT COUNT(*) FROM tpch_dev.bronze.part), (SELECT COUNT(*) FROM samples.tpch.part)
UNION ALL
SELECT 'partsupp', (SELECT COUNT(*) FROM tpch_dev.bronze.partsupp), (SELECT COUNT(*) FROM samples.tpch.partsupp)
UNION ALL
SELECT 'region', (SELECT COUNT(*) FROM tpch_dev.bronze.region), (SELECT COUNT(*) FROM samples.tpch.region)
UNION ALL
SELECT 'supplier', (SELECT COUNT(*) FROM tpch_dev.bronze.supplier), (SELECT COUNT(*) FROM samples.tpch.supplier)
ORDER BY table_name
```
